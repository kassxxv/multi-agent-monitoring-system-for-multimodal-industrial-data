```python
turbine_df = pd.read_excel("../data/edp-wind-scada/T01_F99.xlsx")
turbine_df["Timestamp"] = pd.to_datetime(turbine_df["Timestamp"])
turbine_df = turbine_df.sort_values("Timestamp").reset_index(drop=True)

print("sorted:", turbine_df["Timestamp"].is_monotonic_increasing)
print("missing values total:", turbine_df.isna().sum().sum())
print("duplicate timestamps:", turbine_df.duplicated(subset="Timestamp").sum())
print(turbine_df["Timestamp"].diff().value_counts().head())  # expected sampling interval
```

```python
# one representative column per physical quantity.
signal_columns = [
    "Timestamp",
    "Gen_RPM_Avg",
    "Gen_Bear_Temp_Avg",
    "Gear_Oil_Temp_Avg",
    "Gear_Bear_Temp_Avg",
    "Nac_Temp_Avg",
    "Rtr_RPM_Avg",
    "Amb_WindSpeed_Avg",
    "Blds_PitchAngle_Avg",
    "Grd_Prod_Pwr_Avg",
]
signals_df = turbine_df[signal_columns]
signals_df.describe()
```

```python
logs_2016 = pd.read_excel("../data/edp-wind-scada/Wind-Turbines-Logs-2016.xlsx")
logs_2017 = pd.read_excel("../data/edp-wind-scada/Wind Turbines Logs 2017.xlsx")

# 2016 and 2017 use different column names for the same fields
logs_2017 = logs_2017.rename(columns={
    "Time_Detected": "TimeDetected",
    "Time_Reset": "TimeReset",
    "Turbine_Identifier": "UnitTitle",
    "Unit_Title_Destination": "UnitTitleDestination",
})

logs_df = pd.concat([logs_2016, logs_2017], ignore_index=True)
logs_df["TimeDetected"] = pd.to_datetime(logs_df["TimeDetected"])

# A few source rows have no detection time at all
logs_df = logs_df.dropna(subset=["TimeDetected"])

t01_logs = logs_df[logs_df["UnitTitle"] == "T01"].sort_values("TimeDetected")
print("T01 log events:", t01_logs.shape[0])
t01_logs.head(10)
```

```python
failures_df = pd.read_excel("../data/edp-wind-scada/Combined-Failure-Logbook-2016-2017.xlsx")
t01_failures = failures_df[failures_df["Turbine_ID"] == "T01"]
t01_failures
```

```python
# merge_asof lines up each log event with the nearest SCADA reading in time
# this is how the aggregator will later join sensor state and log state.
aligned = pd.merge_asof(
    t01_logs.sort_values("TimeDetected"),
    signals_df.sort_values("Timestamp"),
    left_on="TimeDetected",
    right_on="Timestamp",
    direction="nearest",
)
aligned.head(10)
```

Below: T01 gearbox pump failure (logged 2016-07-18 02:10)

```python
failure_time = pd.Timestamp("2016-07-18 02:10:00", tz="UTC")
window = pd.Timedelta(days=7)

window_df = turbine_df[
    (turbine_df["Timestamp"] >= failure_time - window)
    & (turbine_df["Timestamp"] <= failure_time + window)
]

fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

axes[0].plot(window_df["Timestamp"], window_df["Gear_Oil_Temp_Avg"], label="Gear_Oil_Temp_Avg")
axes[0].plot(window_df["Timestamp"], window_df["Gear_Bear_Temp_Avg"], label="Gear_Bear_Temp_Avg")
axes[0].set_ylabel("Temp (°C)")
axes[0].legend()

axes[1].plot(window_df["Timestamp"], window_df["Rtr_RPM_Avg"], color="green", label="Rtr_RPM_Avg")
axes[1].set_ylabel("Rotor RPM")
axes[1].legend()

axes[2].plot(window_df["Timestamp"], window_df["Grd_Prod_Pwr_Avg"], color="orange", label="Grd_Prod_Pwr_Avg")
axes[2].set_ylabel("Power (kW)")
axes[2].legend()

for ax in axes:
    ax.axvline(failure_time, color="red", linestyle="--")

axes[0].set_title("T01 gearbox-related signals: 7 days before/after confirmed failure (2016-07-18)")
plt.xlabel("Time")
plt.tight_layout()
plt.savefig("../doc/t01_gearbox_failure_window.png", dpi=100)
plt.show()
```
